```python
import matplotlib.pyplot as plt
import chipwhisperer as cw
import numpy as np
import time 
scope = cw.scope()
scope.default_setup()
target = cw.target(scope, cw.targets.SimpleSerial2)
target.baud = 9600   


traces_good, traces_bad = [], []

for i in range(200):
    cmd = b"f" + bytes([np.random.randint(0,256)]) + b"\n" 
    time.sleep(0.1) 
    scope.arm()
    target.ser.write(cmd)
    if not scope.capture():
        traces_good.append(scope.get_last_trace())

for i in range(200):
    cmd = bytes([np.random.randint(0,256)]) + b"\n"  
    time.sleep(0.1)
    scope.arm()
    target.ser.write(cmd)
    if not scope.capture():
        traces_bad.append(scope.get_last_trace())


plt.figure(figsize=(12,6))  
plt.plot(np.mean(traces_good, axis=0), label="1er octet correct")   
plt.plot(np.mean(traces_bad, axis=0), label="1er octet faux")
plt.xlim(1600, 1900)         
plt.legend()                 
plt.show()  
```
