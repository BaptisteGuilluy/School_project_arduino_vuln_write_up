```python
import chipwhisperer as cw

# 1. Initialisation

scope = cw.scope()
scope.default_setup()
target = cw.target(scope, cw.targets.SimpleSerial2)
target.baud = 9600

def get_synchronized_trace(cmd=b"coucou\n"):
    scope.arm()            
    target.ser.write(cmd)   
    

    if scope.capture():
        print("[-] Timeout : Trigger non détecté")
        return None
        
    return scope.get_last_trace() 


trace = get_synchronized_trace()
if trace is not None:
    print("[+] Trace capturée et synchronisée avec succès !") 

#!pip install "zarr<3"

```
