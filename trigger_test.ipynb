```python
import chipwhisperer as cw
import numpy as np
import time


traces = []
textins = []

number_of_traces = 5000
print(f"[*] Début de la collecte de {number_of_traces} traces pour CPA...")

for i in range(number_of_traces):
    
    textin = np.random.bytes(16) # Pour de l'AES par exemple (16 octets)
    cmd = b"encrypt_" + textin + b"\n"
    
    scope.arm()
    target.ser.write(cmd)
    
    if scope.capture():
        continue # Ignore si timeout
        
    trace = scope.get_last_trace()
    
    traces.append(trace)
    textins.append(textin)
    
    if i % 100 == 0:
        print(f"[*] {i}/{number_of_traces} traces collectées...")


traces = np.array(traces)
textins = np.array(textins)
print("[+] Collecte terminée avec succès !")

```

```python
import numpy as np
import matplotlib.pyplot as plt


sbox = np.array([
    0x63, 0x7c, 0x77, 0x7b, 0xf2, 0x6b, 0x6f, 0xc5, 0x30, 0x01, 0x67, 0x2b, 0xfe, 0xd7, 0xab, 0x76,
    0xca, 0x82, 0xc9, 0x7d, 0xfa, 0x59, 0x47, 0xf0, 0xad, 0xd4, 0xa2, 0xaf, 0x9c, 0xa4, 0x72, 0xc0,
    0xb7, 0xfd, 0x93, 0x26, 0x36, 0x3f, 0xf7, 0xcc, 0x34, 0xa5, 0xe5, 0xf1, 0x71, 0xd8, 0x31, 0x15,
    0x04, 0xc7, 0x23, 0xc3, 0x18, 0x96, 0x05, 0x9a, 0x07, 0x12, 0x80, 0xe2, 0xeb, 0x27, 0xb2, 0x75,
    0x09, 0x83, 0x2c, 0x1a, 0x1b, 0x6e, 0x5a, 0xa0, 0x52, 0x3b, 0xd6, 0xb3, 0x29, 0xe3, 0x2f, 0x84,
    0x53, 0xd1, 0x00, 0xed, 0x20, 0xfc, 0xb1, 0x5b, 0x6a, 0xcb, 0xbe, 0x39, 0x4a, 0x4c, 0x58, 0xcf,
    0xd0, 0xef, 0xaa, 0xfb, 0x43, 0x4d, 0x33, 0x85, 0x45, 0xf9, 0x02, 0x7f, 0x50, 0x3c, 0x9f, 0xa8,
    0x51, 0xa3, 0x40, 0x8f, 0x92, 0x9d, 0x38, 0xf5, 0xbc, 0xb6, 0xda, 0x21, 0x10, 0xff, 0xf3, 0xd2,
    0xcd, 0x0c, 0x13, 0xec, 0x5f, 0x97, 0x44, 0x17, 0xc4, 0xa7, 0x7e, 0x3d, 0x64, 0x5d, 0x19, 0x73,
    0x60, 0x81, 0x4f, 0xdc, 0x22, 0x2a, 0x90, 0x88, 0x46, 0xee, 0xb8, 0x14, 0xde, 0x5e, 0x0b, 0xdb,
    0xe0, 0x32, 0x3a, 0x0a, 0x49, 0x06, 0x24, 0x5c, 0xc2, 0xd3, 0xac, 0x62, 0x91, 0x95, 0xe4, 0x79,
    0xe7, 0xc8, 0x37, 0x6d, 0x8d, 0xd5, 0x4e, 0xa9, 0x6c, 0x56, 0xf4, 0xea, 0x65, 0x7a, 0xae, 0x08,
    0xba, 0x78, 0x25, 0x2e, 0x1c, 0xa6, 0xb4, 0xc6, 0xe8, 0xdd, 0x74, 0x1f, 0x4b, 0xbd, 0x8b, 0x8a,
    0x70, 0x3e, 0xb5, 0x66, 0x48, 0x03, 0xf6, 0x0e, 0x61, 0x35, 0x57, 0xb9, 0x86, 0xc1, 0x1d, 0x9e,
    0xe1, 0xf8, 0x98, 0x11, 0x69, 0xd9, 0x8e, 0x94, 0x9b, 0x1e, 0x87, 0xe9, 0xce, 0x55, 0x28, 0xdf,
    0x8c, 0xa1, 0x89, 0x0d, 0xbf, 0xe6, 0x42, 0x68, 0x41, 0x99, 0x2d, 0x0f, 0xb0, 0x54, 0xbb, 0x16,
], dtype=np.uint8)


hw_table = np.array([bin(i).count("1") for i in range(256)], dtype=np.float32)

KNOWN_NIBBLES = {0: 0xF}


def candidate_range(byte_idx):
    if byte_idx in KNOWN_NIBBLES:
        high = KNOWN_NIBBLES[byte_idx]
        return range(high * 16, high * 16 + 16)
    return range(256)


valid_traces, valid_textins = [], []
for t, p in zip(traces, textins):
    if isinstance(p, (bytes, bytearray)) and len(p) == 16:
        valid_traces.append(t)
        valid_textins.append(list(p))

traces_np = np.array(valid_traces, dtype=np.float32)
textins_np = np.array(valid_textins, dtype=np.uint8)

n_traces, n_points = traces_np.shape
print(f"[*] Lancement de l'attaque CPA sur {n_traces} traces ({n_points} points/trace)...")
if KNOWN_NIBBLES:
    details = ", ".join(f"octet {b} commence par {hex(v)[2:]}" for b, v in KNOWN_NIBBLES.items())
    print(f"[*] Indice(s) pris en compte : {details}")

# ----------------------------------------------------------------------
# 4. Attaque CPA octet par octet (16 octets de l'AES)
# ----------------------------------------------------------------------
extracted_key = []
confidence = []                 # écart meilleur / second meilleur, par octet
all_corr_curves = {}            # corr_curves[byte_idx][k] = courbe de corrélation (pour tracer)

t_centered = traces_np - np.mean(traces_np, axis=0, keepdims=True)
t_std = np.std(traces_np, axis=0)
t_std[t_std == 0] = 1.0

for byte_idx in range(16):
    pt_byte = textins_np[:, byte_idx]
    scores = {}       # k -> corrélation max (sur tout le temps)
    curves = {}        # k -> courbe de corrélation complète (pour les meilleurs)

    for k in candidate_range(byte_idx):
        intermediate = sbox[pt_byte ^ k]
        hyp = hw_table[intermediate]

        h_centered = hyp - np.mean(hyp)
        h_std = np.std(hyp)
        if h_std == 0:
            continue

        corr = np.dot(h_centered, t_centered) / (n_traces * h_std * t_std)
        scores[k] = np.max(np.abs(corr))
        curves[k] = corr

    # Meilleur candidat et second meilleur, pour évaluer la confiance
    ranked = sorted(scores.items(), key=lambda kv: kv[1], reverse=True)
    best_k, best_score = ranked[0]
    second_score = ranked[1][1] if len(ranked) > 1 else 0.0

    extracted_key.append(best_k)
    confidence.append(best_score - second_score)
    all_corr_curves[byte_idx] = (best_k, curves[best_k], ranked[:5])

    print(f"  Octet {byte_idx:2d} : clé = 0x{best_k:02x}  "
          f"(corr max = {best_score:.3f}, écart avec 2e candidat = {best_score - second_score:.3f})")

print("\n[+] Attaque CPA terminée.")
print("Clé trouvée (hex) :", " ".join(f"{b:02x}" for b in extracted_key))

# Alerte si un octet a un résultat peu fiable (second candidat trop proche)
faibles = [i for i, c in enumerate(confidence) if c < 0.05]
if faibles:
    print(f"[!] Octets avec une confiance faible (à vérifier) : {faibles}")

# ----------------------------------------------------------------------
# 5. Affichage graphique
# ----------------------------------------------------------------------

# 5a. Quelques traces de puissance brutes
plt.figure(figsize=(10, 4))
n_show = min(20, n_traces)
for i in range(n_show):
    plt.plot(traces_np[i], linewidth=0.6, alpha=0.6)
plt.title(f"Traces de puissance (échantillon de {n_show}/{n_traces})")
plt.xlabel("Point temporel")
plt.ylabel("Amplitude")
plt.tight_layout()
# plt.savefig("/outputs/traces_brutes.png", dpi=150)
plt.close()

# 5b. Courbes de corrélation pour chaque octet (meilleur candidat)
fig, axes = plt.subplots(4, 4, figsize=(16, 10), sharex=True)
for byte_idx, ax in enumerate(axes.flat):
    best_k, curve, _ = all_corr_curves[byte_idx]
    ax.plot(curve, linewidth=0.8, color="steelblue")
    peak_idx = np.argmax(np.abs(curve))
    ax.axvline(peak_idx, color="red", linestyle="--", linewidth=0.8)
    ax.set_title(f"Octet {byte_idx} -> 0x{best_k:02x}", fontsize=9)
    ax.tick_params(labelsize=7)
fig.suptitle("Corrélation en fonction du temps (meilleur candidat par octet)")
fig.tight_layout()
# fig.savefig("/mnt/user-data/outputs/correlations_par_octet.png", dpi=150)
plt.close(fig)

# 5c. Top 5 candidats de l'octet 0 (celui pour lequel on a un indice),
#     pour visualiser à quel point le bon candidat se détache des autres.
best_k0, _, top5 = all_corr_curves[0]
pt_byte0 = textins_np[:, 0]
plt.figure(figsize=(10, 4))
for k, score in top5:
    intermediate = sbox[pt_byte0 ^ k]
    hyp = hw_table[intermediate]
    h_centered = hyp - np.mean(hyp)
    h_std = np.std(hyp)
    corr = np.dot(h_centered, t_centered) / (n_traces * h_std * t_std)
    style = "-" if k == best_k0 else "--"
    width = 1.4 if k == best_k0 else 0.7
    plt.plot(corr, style, linewidth=width, label=f"0x{k:02x} (max={score:.3f})")
plt.legend(fontsize=8)
plt.title("Octet 0 : top 5 candidats (recherche restreinte à 0xF0-0xFF)")
plt.xlabel("Point temporel")
plt.ylabel("Corrélation")
plt.tight_layout()
# plt.savefig("/mnt/user-data/outputs/top5_octet0.png", dpi=150)
plt.close()

print("\n[+] Graphiques enregistrés :")
print("   - traces_brutes.png")
print("   - correlations_par_octet.png")
print("   - top5_octet0.png")
```

```python
 
import chipwhisperer as cw
import time
 
key_hex_ascii = "".join(f"{b:02x}" for b in extracted_key)  # ex: "f31a..."
 
print("\n[*] Clé à tester (ASCII) :", key_hex_ascii)
 
scope = cw.scope()
target = cw.target(scope, cw.targets.SimpleSerial2)
target.baud = 9600
 
 
scope.io.nrst = "low"
time.sleep(0.05)
scope.io.nrst = "high"
time.sleep(0.05)
target.ser.read(200) #vide le buffer


 
def lire_tout(timeout=1.0, pas=0.05):
    out = ""
    silence = 0.0
    while silence < timeout:
        chunk = target.ser.read(50)
        if chunk:
            out += chunk
            silence = 0.0
        else:
            time.sleep(pas)
            silence += pas
    return out
 

boot = lire_tout(timeout=0.5)
print("Boot message:", repr(boot))

print("\n[*] Test diagnostic avec un mauvais mot de passe court (\"1234\") :")
target.ser.write("1234\n")
rep_test = lire_tout()
print("Response (test):", repr(rep_test))
 
# --- Envoi de la clé trouvée ---
print(f"\n[*] Envoi du mot de passe : {key_hex_ascii} (longueur={len(key_hex_ascii)})")
 
scope.io.nrst = "low"
time.sleep(0.05)
scope.io.nrst = "high"
time.sleep(0.05)
target.ser.read(200) #vide le buffer

target.ser.write(key_hex_ascii + "\n")
reponse = lire_tout()
print("Response:", repr(reponse))
 
MOTS_ECHEC = ("denied", "incorrect", "wrong", "error")
MOTS_SUCCES = ("granted", "welcome", "unlock")
 
succes_initial = reponse and not any(m in reponse.lower() for m in MOTS_ECHEC)
succes_initial = succes_initial or any(m in reponse.lower() for m in MOTS_SUCCES)
 
if succes_initial:
    print("\n[+] La cible semble avoir accepté la clé !")
else:
    print("\n[!] Candidat #1 refusé (ACCESS DENIED). Passage à l'automatisation "
          "pour tester les candidats suivants.")

def tester_cle(cle_bytes, description=""):

    hexa = "".join(f"{b:02x}" for b in cle_bytes)
    scope.io.nrst = "low"
    time.sleep(0.05)
    scope.io.nrst = "high"
    time.sleep(0.05)
    target.ser.read(200) #vide le buffer
    target.ser.write(hexa + "\n")
    rep = lire_tout()
    succes = rep and not any(m in rep.lower() for m in MOTS_ECHEC)
    succes = succes or any(m in rep.lower() for m in MOTS_SUCCES)
    print(f"[{'OK ' if succes else '...'}] {description or hexa} -> {rep!r}")
    return succes, rep
 
 
meilleure_cle = list(extracted_key)
succes = succes_initial
 
if not succes:
    # Octets triés du moins fiable (écart faible = ambigu) au plus fiable
    ordre_octets = sorted(range(16), key=lambda i: confidence[i])
    print(f"[*] Ordre d'exploration des octets (du moins au plus fiable) : {ordre_octets}")
 
    for byte_idx in ordre_octets:
        _, _, candidats = all_corr_curves[byte_idx]  # [(k, score), ...] triés
        for k, score in candidats[1:]:  # on saute le candidat déjà testé
            essai = list(meilleure_cle)
            essai[byte_idx] = k
            succes, rep = tester_cle(
                essai, f"octet {byte_idx}: 0x{meilleure_cle[byte_idx]:02x} -> 0x{k:02x} (corr={score:.3f})"
            )
            if succes:
                meilleure_cle = essai
                break
        if succes:
            break
 
if succes:
    print("\n[+] Mot de passe trouvé :", "".join(f"{b:02x}" for b in meilleure_cle))
else:
    print("\n[!] Aucun candidat (parmi les top 5 par octet) n'a fonctionné. "
          "Pistes : capturer plus de traces, vérifier le point d'intérêt "
          "temporel, ou essayer un autre modèle de fuite (HD au lieu de HW, "
          "par exemple).")
 
scope.dis()
```
